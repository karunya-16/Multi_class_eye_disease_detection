```python
# Packages
import os
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Dataset Path
dataset_path = "../dataset/Messidor-2+EyePac_Balanced"

classes = ["0", "1", "2", "3", "4"]

image_extensions = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
)

print(os.path.abspath(dataset_path))
```

```python
# All images List
data = []

for cls in classes:

    folder = os.path.join(dataset_path, cls)

    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith(image_extensions)
    ]

    for file in files:

        image_path = os.path.join(folder, file)

        data.append({
            "image_path": image_path,
            "label": int(cls)
        })

df = pd.DataFrame(data)

print("Total images:", len(df))
print("\nClass distribution:")
print(df["label"].value_counts().sort_index())
```

```python
# First split: Train + Temporary
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["label"],
    random_state=42
)

print("Training images:", len(train_df))
print("Remaining images:", len(temp_df))
```

```python
# Temporary → Validation + Test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=42
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
# Check class distribution
print("TRAIN DISTRIBUTION")
print(train_df["label"].value_counts().sort_index())

print("\nVALIDATION DISTRIBUTION")
print(val_df["label"].value_counts().sort_index())

print("\nTEST DISTRIBUTION")
print(test_df["label"].value_counts().sort_index())
```

```python
os.makedirs("../preprocessing/splits", exist_ok=True)

train_df.to_csv(
    "../preprocessing/splits/train.csv",
    index=False
)

val_df.to_csv(
    "../preprocessing/splits/validation.csv",
    index=False
)

test_df.to_csv(
    "../preprocessing/splits/test.csv",
    index=False
)

print("Split CSV files saved successfully!")
```
