```python
import os

dataset_path = "../dataset/Messidor-2+EyePac_Balanced"

classes = ["0", "1", "2", "3", "4"]

for cls in classes:
    folder = os.path.join(dataset_path, cls)

    if os.path.exists(folder):
        files = os.listdir(folder)

        print(f"Class {cls}: {len(files)} images")
    else:
        print(f"Class {cls}: Folder not found")
```

```python
from PIL import Image
import os

for cls in classes:

    folder = os.path.join(dataset_path, cls)

    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"))
    ]

    if len(files) > 0:

        image_path = os.path.join(folder, files[0])

        image = Image.open(image_path)

        print(f"Class {cls}")
        print("Image:", files[0])
        print("Size:", image.size)
        print("Format:", image.format)
        print()
```

```python
from PIL import Image
import os

for cls in classes:
    folder = os.path.join(dataset_path, cls)

    files = os.listdir(folder)

    if len(files) > 0:
        image_path = os.path.join(folder, files[0])

        try:
            img = Image.open(image_path)

            print(f"Class {cls}")
            print("Image:", files[0])
            print("Size:", img.size)
            print("Format:", img.format)
            print("-" * 40)

        except Exception as e:
            print("Error:", image_path, e)
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os

plt.figure(figsize=(15, 8))

for i, cls in enumerate(classes):

    folder = os.path.join(dataset_path, cls)
    files = os.listdir(folder)

    image_path = os.path.join(folder, files[0])

    img = Image.open(image_path)

    plt.subplot(1, 5, i + 1)
    plt.imshow(img)
    plt.title(f"Class {cls}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = []

for cls in classes:
    folder = os.path.join(dataset_path, cls)
    counts.append(len(os.listdir(folder)))

plt.figure(figsize=(8, 5))
plt.bar(classes, counts)
plt.xlabel("Disease Severity Class")
plt.ylabel("Number of Images")
plt.title("Dataset Class Distribution")
plt.show()

print(dict(zip(classes, counts)))
```
